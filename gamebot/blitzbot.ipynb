```python
import pyautogui as pyg
pyg.sleep(2)
basex=65
basey=470
gap=40
```

```python
im = pyg.screenshot()
for row in range(8):
    for col in range(8):
        x = basex + col * gap
        y = basey + row * gap
        color = im.getpixel((x, y))
        print(f'({row}, {col}) at ({x}, {y}): {color}')
```
