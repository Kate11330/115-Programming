```python
import pyautogui as pyg
pyg.sleep(2)
basex=81
basey=346
gap=40
```

```python
def gem_position(r, c):
    """回傳第 r 列、第 c 欄寶石中心點的 (x, y) 座標；row、column 從 0 開始。"""
    return (basex + c * gap, basey + r * gap)
```

```python
def get_gem_color(r, c, im):
    """回傳截圖 im 中第 r 列、第 c 欄寶石中心點的顏色。"""
    x, y = gem_position(r, c)
    return im.getpixel((x, y))
```

```python
def can_move_left(r, c, im):
    """判斷將 (r, c) 的寶石與左側寶石交換後，是否會形成三連線。"""
    if not (0 <= r < 8 and 1 <= c < 8):
        return False

    left_c = c - 1
    current_color = get_gem_color(r, c, im)
    left_color = get_gem_color(r, left_c, im)

    def color_after_swap(row, col):
        if (row, col) == (r, c):
            return left_color
        if (row, col) == (r, left_c):
            return current_color
        return get_gem_color(row, col, im)

    # 只需檢查交換後兩個位置，因為新形成的連線必定經過其中之一。
    for row, col in ((r, c), (r, left_c)):
        color = color_after_swap(row, col)
        for dr, dc in ((0, 1), (1, 0)):
            count = 1
            for direction in (-1, 1):
                nr, nc = row + direction * dr, col + direction * dc
                while 0 <= nr < 8 and 0 <= nc < 8 and color_after_swap(nr, nc) == color:
                    count += 1
                    nr += direction * dr
                    nc += direction * dc
            if count >= 3:
                return True

    return False
```

```python
def _can_move_in_direction(r, c, im, dr, dc):
    """判斷 (r, c) 與指定方向的相鄰寶石交換後是否會形成三連線。"""
    nr, nc = r + dr, c + dc
    if not (0 <= r < 8 and 0 <= c < 8 and 0 <= nr < 8 and 0 <= nc < 8):
        return False

    current_color = get_gem_color(r, c, im)
    neighbor_color = get_gem_color(nr, nc, im)

    def color_after_swap(row, col):
        if (row, col) == (r, c):
            return neighbor_color
        if (row, col) == (nr, nc):
            return current_color
        return get_gem_color(row, col, im)

    # 交換後的新連線必定經過其中一個交換位置。
    for row, col in ((r, c), (nr, nc)):
        color = color_after_swap(row, col)
        for line_dr, line_dc in ((0, 1), (1, 0)):
            count = 1
            for direction in (-1, 1):
                check_r = row + direction * line_dr
                check_c = col + direction * line_dc
                while (0 <= check_r < 8 and 0 <= check_c < 8
                       and color_after_swap(check_r, check_c) == color):
                    count += 1
                    check_r += direction * line_dr
                    check_c += direction * line_dc
            if count >= 3:
                return True

    return False


def can_move_right(r, c, im):
    return _can_move_in_direction(r, c, im, 0, 1)


def can_move_can_right(r, c, im):
    """can_move_right 的別名。"""
    return can_move_right(r, c, im)


def can_move_up(r, c, im):
    return _can_move_in_direction(r, c, im, -1, 0)


def can_move_down(r, c, im):
    return _can_move_in_direction(r, c, im, 1, 0)
```
