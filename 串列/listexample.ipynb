```python
# scores是一個串列，裡面每個元素都是整數。
scores=[78,29,56,80,99]
print(scores)
```

我們可以使用索引值(index)來存取串列中的元素。

注意：索引值由0開始

```python
print(scores[0]) # 印出第1個元素
print(scores[1]) # 印出第2個元素
```

## 串列內容的個數

len(a)取得a這個串列中的個數。
len是length(長度)的意思

```python
print(len(scores))
print(scores[4]) # 取得最後一個元素
print(scores[5-1]) # 取得最後一個元素
print(scores[-1])
```
